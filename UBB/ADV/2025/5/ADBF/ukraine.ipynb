```python
df2=df[['event_id_cnty', 'event_date', 'event_type', 'sub_event_type', 'location', 'latitude', 'longitude',
       'notes', 'fatalities','timestamp']]
```

```python
# Convert Unix timestamp to datetime
df2['datetime']=pd.to_datetime(df2['timestamp'], unit='s')
DURATION='12h'
df2['datetime2']=df2['datetime']+pd.Timedelta(DURATION)

```

```python
df2.to_csv('Ukraine_clean2.csv',index=False)
```

```python
df2.to_csv('Ukraine_clean.csv',index=False)
```

```python
df3=df_full[['event_id_cnty', 'event_date', 'event_type', 'sub_event_type', 'location', 'latitude', 'longitude',
       'notes', 'fatalities','timestamp']]

# Convert Unix timestamp to datetime
df3['datetime']=pd.to_datetime(df3['timestamp'], unit='s')
DURATION='12h'
df3['datetime2']=df3['datetime']+pd.Timedelta(DURATION)

df3.to_csv('Ukraine_clean_full.csv',index=False)

```
